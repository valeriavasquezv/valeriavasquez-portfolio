```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/valeriavasquezv/valeriavasquez-portfolio/refs/heads/main/Proyecto_TelecomX/TelecomX_Data.json"
```

## EXTRACCION DE LOS DATOS

```python
df_telecom = pd.read_json(url)
df_telecom
```

```python
df_telecom.info()
```

```python
df_telecom.dtypes
```

##TRANSFORMACIÓN DE LOS DATOS


```python
churn_df = df_telecom['Churn']
pd.unique(churn_df)

#columna churn contiene valores en blanco, revisar en siguiente etapa
```

```python
customer_df = pd.json_normalize(df_telecom["customer"])

pd.unique(customer_df["gender"])
pd.unique(customer_df["Partner"])
pd.unique(customer_df["Dependents"])
pd.unique(customer_df["SeniorCitizen"])
pd.unique(customer_df['tenure'])
```

```python
phone_df = pd.json_normalize(df_telecom['phone'])

pd.unique(phone_df['PhoneService'])
pd.unique(phone_df['MultipleLines'])
```

```python
internet_df = pd.json_normalize(df_telecom['internet'])

pd.unique(internet_df['InternetService'])
pd.unique(internet_df['OnlineSecurity'])
pd.unique(internet_df['OnlineBackup'])
pd.unique(internet_df['DeviceProtection'])
pd.unique(internet_df['TechSupport'])
pd.unique(internet_df['StreamingTV'])
pd.unique(internet_df['StreamingMovies'])

```

```python
account_df = pd.json_normalize(df_telecom['account'])

pd.unique(account_df['Contract'])
pd.unique(account_df['PaperlessBilling'])
pd.unique(account_df['PaymentMethod'])
pd.unique(account_df['Charges.Monthly'])
pd.unique(account_df['Charges.Total'])

# Se detectó que las variables mensuales y totales se encuentran almacenadas como texto.
```

### Columna Churn

```python
import numpy as np
```

```python
df_telecom['Churn'] = df_telecom['Churn'].replace('', np.nan)
```

```python
df_telecom = df_telecom.dropna(subset=['Churn'])
pd.unique(df_telecom['Churn'])
```

### Columna Account

```python
account_df['Charges.Monthly'] = pd.to_numeric(
    account_df['Charges.Monthly'], errors='coerce'
)

account_df['Charges.Total'] = pd.to_numeric(
    account_df['Charges.Total'], errors='coerce'
)

account_df.dtypes
```

```python
df_final = pd.concat(
    [df_telecom.drop(columns=['account']), account_df],
    axis=1
)

df_final.columns
```

#CARGA Y ANÁLISIS

```python
df_final[['Charges.Monthly', 'Charges.Total']].describe()
```

```python
df_final.groupby('Churn')[['Charges.Monthly', 'Charges.Total']].mean()
```

```python
import matplotlib.pyplot as plt

churn_counts = df_final['Churn'].value_counts()

plt.pie(
    churn_counts,
    labels=churn_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Distribución de Churn')
plt.show()
```

```python
def churn_by_category(df, column):
    table = pd.crosstab(df[column], df['Churn'], normalize='index')
    display(table)

    table.plot(kind='bar', stacked=True)
    plt.title(f'Distribución de Churn según {column}')
    plt.ylabel('Proporción')
    plt.xlabel(column)
    plt.legend(title='Churn')
    plt.show()
```

```python
# Tipo de contrato

churn_by_category(df_final, 'Contract')
```

```python
# Método de pago

churn_by_category(df_final, 'PaymentMethod')
```

```python
# facturación electrónica

churn_by_category(df_final, 'PaperlessBilling')
```

```python
# Gasto mensual vs Churn

df_final.boxplot(column='Charges.Monthly', by='Churn')
plt.title('Distribución del cargo mensual según Churn')
plt.suptitle('')
plt.ylabel('Charges.Monthly')
plt.show()
```

```python
# Gasto total vs Churn

df_final.boxplot(column='Charges.Total', by='Churn')
plt.title('Distribución del gasto total según Churn')
plt.suptitle('')
plt.ylabel('Charges.Total')
plt.show()
```

## 1. Introducción

El objetivo de este desafío es implementar un **proceso ETL (Extract, Transform, Load)** sobre un conjunto de datos de clientes de telecomunicaciones, asegurando la correcta **extracción, limpieza, transformación y validación de los datos**.

Como parte de la validación del proceso, se realiza un **análisis exploratorio básico** de la variable **Churn (evasión de clientes)**, con el fin de comprobar la coherencia de los datos y obtener insights iniciales que permitan verificar la calidad de la información procesada.

---

## 3. Transform – Limpieza y Tratamiento de Datos

Durante la etapa de transformación se realizaron los siguientes pasos:

* Corrección de **tipos de datos**, especialmente en:

  * `Charges.Total`, originalmente almacenada como texto.
* Conversión de variables numéricas a formato adecuado para análisis.
* Eliminación o tratamiento de valores faltantes e inconsistentes.
* Normalización de categorías en variables como:

  * `Contract`
  * `PaymentMethod`
  * `PaperlessBilling`
* Validación de la variable `Churn`, asegurando consistencia entre sus categorías.

Estas transformaciones aseguran que los datos sean **confiables, consistentes y listos para análisis o modelado posterior**.

## 5. Validación del ETL – Análisis Exploratorio

### 5.1 Distribución de la variable Churn

* Aproximadamente **73.5%** de los clientes permanecen activos.
* Cerca de **26.5%** presentan evasión.

Este análisis confirma una distribución coherente de la variable objetivo tras el proceso ETL.

---

### 5.2 Análisis por variables categóricas

* **Tipo de contrato:**
  Los contratos mes a mes presentan mayor evasión, mientras que los contratos anuales muestran mayor retención.

* **Método de pago:**
  El método *Electronic check* presenta la mayor tasa de churn, mientras que los pagos automáticos muestran menor evasión.

* **Facturación electrónica:**
  Los clientes con facturación electrónica presentan una mayor proporción de evasión.

Estos resultados validan que las transformaciones aplicadas mantienen relaciones lógicas entre variables.

---

### 5.3 Análisis por variables numéricas

* **Cargos mensuales:**
  Los clientes que cancelan tienden a presentar cargos mensuales más altos.

* **Gasto total:**
  Los clientes que permanecen activos acumulan un gasto total mayor, reflejando mayor antigüedad.

Este análisis confirma la coherencia de los datos numéricos tras la limpieza y transformación.
